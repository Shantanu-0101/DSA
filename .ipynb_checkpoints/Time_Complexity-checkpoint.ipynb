## Counting Operation

```python
def c_to_f(c):
    return c*9.0/5 + 32  #Total 3 operations
```

```python
def mysum(x):
    total=0  # 1 operation
    for i in range(x+1):   # loop x times
        total += 1   # 2 operation x loop times
    return total
```

# Order of Growth (most efficient)
### Big O notation 
### O(n)
* computes factorial
* number of steps
* worst-case asymptotic complexity:
  - ignore additive constants
  - ignore multiplicative constants

```python
def fact_iter(n):
    # assumes n an int >= 0
    answer = 1
    while n > 1:
        answer *= n
        n -=1
    return answer
```

* Types of order of growth
  - constant
  - linear
  - quadratic
  - logarithmic
  - n log n
  - exponential

```python
# Problem 1

L =  [1,2,3,4,5]

sum = 0
for i in L:
    sum = sum+1
print(sum)


product = 1
for i in L:
    product = product*1
print(product)

# Answer: Two independent loop
#          O(n) + O(n) = O(n+n)
#                      = O(2n)
#                      = O(n)  Linear Time Complexity
```

```python
# Problem 2

L = [1,2,3,4,5]

for i in L:
    for j in L:
        print('({}, {})'.format(i,j))

# Answer: Nested loop problem (multiplication)
#         O(n) x O(n) = O(nxn) = O(n^2)
#         Quadratic Time Complexity
```

```python
# Problem 3

def intTOstr(i):
    digits = '0123456789'
    if i == 0:
        return '0'
    result = ''
    while i>0:
        result = digits[i%10] + result
        i = i//10
    return result

intTOstr(2)

# Answer: log
```

```python
# Problem 4

n = 1000
int i,j,k = 0

for(i=n/2;j<=n;i++){
    for(j=2;j<=n;j*2){
        k=k+n/2
    }
}

# Answer: This is the problem of O(nlogn)
#         O(n/2) x O(logn)
#         O(n/2 x logn)
#         O(1/2 nlogn)
#         O(nlogn)
```
